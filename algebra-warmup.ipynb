# Plotting Points

Today's first tasks:
1. Plot a single point
1. Plot multiple points

We'll use [`plt.scatter()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html#matplotlib.pyplot.scatter.) to plot disconnected points.

It can take many arguments, but the first two positional arguments are the most important: `x` and `y`.

## Plot a Single Point

Plot a point at **x=5** and **y=10**

```python
dot_coordinates = [5, 10]

plt.scatter(x=dot_coordinates[0], y=dot_coordinates[1]);
```

## Plot Multiple Points

Plot several points stored in a list with **shape (2, 13)** (i.e., a list of lists).

Hint: the `x` and `y` arguments can be lists of x's and y's.

```python
dots = [
    [1, 1, 2, 2, 2, 3, 1, 3, 3, 3, 3, 0, 0],
    [2, 3, 2, 4, 1, 3, 4, 2, 0, 5, 7, 2, 3]
]

plt.scatter(x=dots[0], y=dots[1]);
```

In this case we used lists for x and y, but this works equally well with NumPy 1D-arrays or Pandas Series.

## Plot a Line

Plot a line starting at **x=4, y=0** and ending at **x=20, y=4**. The function you need is [pyplot.plot()](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot).

`plt.plot()` plots the **points** you give it and draws a **straight line between the points**. So to draw a line, you only need to provide two points and `plt.plot()` will draw the line between them.

Like `plt.scatter()`, `plt.plot()` expects two arguments: a list of x's and a list of y's.

```python
start = {"x": 4, "y": 0}
end = {"x": 20, "y":4}

plt.plot([start["x"], end["x"]], [start["y"], end["y"]]);
```

## Plotting Functions

To plot a mathematical function, we'll always follow this recipe:

1. Define the range of x-values you want to plot
1. Generate a list or array of evenly spaced x-values within that range
1. Compute the corresponding y-values for those x-values using your mathematical function
1. Plot the x- and y-values

### Plot a Function

❓ Generate data points in the range `[-10, 10]`. These are your `x` points. Use them to plot the following function:

$$y = 4x + 2$$

Use [ plt.ylim(-10, 10)](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.ylim.html#matplotlib.pyplot.ylim) to keep the **y-axis** between **-10 and 10**

<br/>

<details>
  <summary>💡 Hint</summary>
  You can use <code><a href="https://docs.scipy.org/doc/numpy/reference/generated/numpy.arange.html">numpy.arange()</a></code> or <code><a href="https://numpy.org/doc/stable/reference/generated/numpy.linspace.html">numpy.linspace()</a></code> to generate a list of N numbers within a range.

  What's the difference?
  - Both take a start and end value
  - With `np.arange()` you define the step size between values
  - With `np.linspace()` you define the number of points between values
</details>

```python
x = np.arange(-10, 10, 20/100)
y = 4*x + 2
plt.plot(x, y)
plt.ylim(-10, 10)
plt.show()
```

### Plot More Advanced Functions

❓ Plot these 4 functions together on a new graph, using the same `x` points generated above, keeping the **y-axis** between **-2 and 6** and the **x-axis** between **-4 and 4**.

$$y1 = x^2$$

$$y2 = e^x$$

$$y3 = \frac{1}{1 + e^{-(-3 + 2*x)}}$$

<details>
  <summary>💡 Hints</summary>
    
  There are multiple ways to do this... to name a few:
  - Use a `for` loop / list comprehension
  - Use NumPy vectorized functions (e.g. `np.exp(your_list_of_x_numbers)`)
  - Use Python's built-in [`map`](https://www.w3schools.com/python/ref_func_map.asp) function

  Pro tip: Use NumPy vectorized functions — write the function as if you were writing math. Not only faster, but also easier to read!

  `np.exp()` will be needed.

</details>

Getting a very angular line? Try with a smaller step size. `plt.plot()` only plots the x's and y's you give it and draws a **straight** line between them. The closer your points are to each other, the smoother your line will be.

```python
x = np.arange(-10, 10, 20/100)

y1 = x**2
y2 = np.exp(x)
y3 = 1/(1 + np.exp(-(-3 + 2*x)))

plt.ylim(-2, 6)
plt.xlim(-4, 4)
plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3);
```

## Combining Points and Lines

Below is a **scatter plot** generated with `sklearn`, a library you'll be learning soon! We want to draw a **red dashed straight line** that follows the spread of the points as closely as possible 🖌️

Our final result should look like this:  
<img src="https://wagon-public-datasets.s3.amazonaws.com/data-science-images/02-Data-Toolkit/03-Data-Visualization/020302-regression-line-example.png" alt="regression_line" width=400>

```python
from sklearn import datasets

x, y, _ = datasets.make_regression(n_samples=100, n_features=1, noise=10, coef=True, random_state=0)

plt.scatter(x,y);
```

❓ No complex calculations needed — just use your eyes to estimate what the line should be.

Draw a red dashed line that **approximately** regresses the scatter plot above
- Reuse the `x` and `y` generated from the previous cell
- Remember: two points are enough to draw a line with `plt.plot()`
- Use these arguments to make your line red and dashed: `color='r', linestyle='--'` <br>(Try other values for `color`: `'g'` or `'c'` or `'m'` and for `linestyle`: `':'` or `'-.'`)

```python
plt.scatter(x,y)
plt.plot([-2, 2], [-95, 88], color='r', linestyle='--')
```

❓ Now, we'll add context to our plot:
1. A title saying "Linear Regression"
1. A legend explaining the data points and the line
1. A label for the x-axis ("input value") and for the y-axis ("output value")

We're providing the code here. Just replace `# YOUR CODE HERE` with the plot you created before, but this time add a `label` to both plots.

```python
# Set the size of the figure
fig = plt.figure(figsize=(10, 5))

# Plot the scatter plot and the accompanying red line
# This time add a `label` to the two plots
plt.scatter(x,y, label='Generated data')
plt.plot([-2, 2], [-95, 88], color='r', linestyle='--', label='Regression line')

# Add title and legend
plt.title("Linear Regression")
plt.legend()

# Add labels for both axes
plt.xlabel("input value")
plt.ylabel("output value")

# Add annotation to the regression line
plt.annotate("Regression line", (0,0), (1,-50), arrowprops={"arrowstyle":"->"});
```
